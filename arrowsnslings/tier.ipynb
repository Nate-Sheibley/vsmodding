```python
# ================================================================
# INTERACTIVE ARMOR vs. BOW / ARROW DAMAGE CALCULATOR
#
# X axis   = Bow type
# Y axis   = Damage taken
# Color    = Armor
# Line     = Projectile
#
# CHECKBOXES:
#   - Select any number of armors
#   - Select any number of projectiles
#   - Every armor × projectile combination is plotted
#
# IMPORTANT:
#   attackTier belongs to the ARROW, not the bow.
#
# Fill in ARROW_ATTACK_TIERS below.
# ================================================================

import pandas as pd
import altair as alt
import ipywidgets as widgets
from IPython.display import display, clear_output


# ================================================================
# ARROW ATTACK TIERS
#
# CHANGE THESE VALUES TO THE CORRECT TIERS.
# ================================================================

BOW_ATTACK_TIERS = {
    "Crude": 0,
    "Simple": 1,
    "Long":2,
    "Cable Backed Long":2,
    "Recurve": 3,
    "Composite": 3
}


# ================================================================
# ARMOR DATA
# ================================================================

armor_stats = {"unarmored": {
        "relProt": 0,
        "flatProt": 0,
        "protTier": -1,
    },
    
    "improvisedwood": {
        "relProt": 0.55,
        "flatProt": 0,
        "protTier": 0,
    },

    "jerkinleather": {
        "relProt": 0.40,
        "flatProt": 0.25,
        "protTier": 1,
    },

    "sewnleather": {
        "relProt": 0.60,
        "flatProt": 0.60,
        "protTier": 1,
        "relProtLossOverTier": 0.015,
        "relProtLossUnderTier": 0.075,
        "flatProtLossOverTier": 0.05,
        "flatProtLossUnderTier": 0.10,
    },

    "sewnlinen": {
        "relProt": 0.75,
        "flatProt": 0.75,
        "protTier": 2,
        "relProtLossOverTier": 0.015,
        "relProtLossUnderTier": 0.075,
        "flatProtLossOverTier": 0.05,
        "flatProtLossUnderTier": 0.10,
    },

    "lamellarwood": {
        "relProt": 0.65,
        "flatProt": 0.50,
        "protTier": 0,
    },

    # "lamellarcopper": {
    #     "relProt": 0.75,
    #     "flatProt": 0.50,
    #     "protTier": 1,
    # },

    "lamellartinbronze": {
        "relProt": 0.77,
        "flatProt": 0.60,
        "protTier": 2,
    },

    "lamellarbismuthbronze": {
        "relProt": 0.76,
        "flatProt": 0.58,
        "protTier": 2,
    },

    # "lamellarblackbronze": {
    #     "relProt": 0.78,
    #     "flatProt": 0.65,
    #     "protTier": 2,
    # },

    # "lamellariron": {
    #     "relProt": 0.79,
    #     "flatProt": 0.70,
    #     "protTier": 3,
    # },

    # "lamellarsteel": {
    #     "relProt": 0.81,
    #     "flatProt": 0.80,
    #     "protTier": 4,
    # },

    # "brigandinecopper": {
    #     "relProt": 0.78,
    #     "flatProt": 1.00,
    #     "protTier": 1,
    # },

    # "brigandinetinbronze": {
    #     "relProt": 0.80,
    #     "flatProt": 1.10,
    #     "protTier": 2,
    # },

    # "brigandinebismuthbronze": {
    #     "relProt": 0.79,
    #     "flatProt": 1.10,
    #     "protTier": 2,
    # },

    # "brigandineblackbronze": {
    #     "relProt": 0.81,
    #     "flatProt": 1.15,
    #     "protTier": 2,
    # },

    "brigandineiron": {
        "relProt": 0.82,
        "flatProt": 1.20,
        "protTier": 3,
    },

    # "brigandinesteel": {
    #     "relProt": 0.84,
    #     "flatProt": 1.30,
    #     "protTier": 4,
    # },

    "chaincopper": {
        "relProt": 0.80,
        "flatProt": 1.10,
        "protTier": 1,
    },

    "chaintinbronze": {
        "relProt": 0.82,
        "flatProt": 1.20,
        "protTier": 2,
    },

    "chainbismuthbronze": {
        "relProt": 0.81,
        "flatProt": 1.20,
        "protTier": 2,
    },

    # "chainblackbronze": {
    #     "relProt": 0.83,
    #     "flatProt": 1.25,
    #     "protTier": 2,
    # },

    "chainiron": {
        "relProt": 0.84,
        "flatProt": 1.30,
        "protTier": 3,
    },

    "chainsteel": {
        "relProt": 0.86,
        "flatProt": 1.40,
        "protTier": 4,
    },

    # "scalecopper": {
    #     "relProt": 0.84,
    #     "flatProt": 1.30,
    #     "protTier": 1,
    # },

    # "scaletinbronze": {
    #     "relProt": 0.86,
    #     "flatProt": 1.35,
    #     "protTier": 2,
    # },

    # "scalebismuthbronze": {
    #     "relProt": 0.85,
    #     "flatProt": 1.30,
    #     "protTier": 2,
    # },

    # "scaleblackbronze": {
    #     "relProt": 0.87,
    #     "flatProt": 1.40,
    #     "protTier": 2,
    # },

    # "scaleiron": {
    #     "relProt": 0.88,
    #     "flatProt": 1.50,
    #     "protTier": 3,
    # },

    # "scalesteel": {
    #     "relProt": 0.90,
    #     "flatProt": 1.60,
    #     "protTier": 4,
    # },

    # "platecopper": {
    #     "relProt": 0.90,
    #     "flatProt": 1.50,
    #     "protTier": 1,
    #     "highDmgTierResistant": True,
    # },

    # "platetinbronze": {
    #     "relProt": 0.92,
    #     "flatProt": 1.60,
    #     "protTier": 2,
    #     "highDmgTierResistant": True,
    # },

    # "platebismuthbronze": {
    #     "relProt": 0.94,
    #     "flatProt": 1.55,
    #     "protTier": 2,
    #     "highDmgTierResistant": True,
    # },

    # "plateblackbronze": {
    #     "relProt": 0.95,
    #     "flatProt": 1.65,
    #     "protTier": 2,
    #     "highDmgTierResistant": True,
    # },

    "plateiron": {
        "relProt": 0.96,
        "flatProt": 1.70,
        "protTier": 3,
        "highDmgTierResistant": True,
    },

    "platesteel": {
        "relProt": 0.97,
        "flatProt": 1.80,
        "protTier": 4,
        "highDmgTierResistant": True,
    },

    # "antiqueblackguardpristine": {
    #     "relProt": 0.96,
    #     "flatProt": 1.70,
    #     "highDmgTierResistant": True,
    #     "protTier": 3,
    # },

    # "antiqueblackguardbroken": {
    #     "relProt": 0.40,
    #     "flatProt": 0.25,
    #     "protTier": 1,
    # },

    # "antiqueblackguarddamaged": {
    #     "relProt": 0.75,
    #     "flatProt": 0.50,
    #     "protTier": 1,
    # },

    # "antiqueforlornpristine": {
    #     "relProt": 0.96,
    #     "flatProt": 1.70,
    #     "highDmgTierResistant": True,
    #     "protTier": 3,
    # },

    # "antiqueforlorndamaged": {
    #     "relProt": 0.90,
    #     "flatProt": 1.50,
    #     "highDmgTierResistant": True,
    #     "protTier": 1,
    # },

    # "antiqueforlornbroken": {
    #     "relProt": 0.40,
    #     "flatProt": 0.25,
    #     "protTier": 1,
    # },
}


# Defaults from the original JavaScript
for armor in armor_stats.values():
    armor.setdefault("relProtLossOverTier", 0.03)
    armor.setdefault("relProtLossUnderTier", 0.15)
    armor.setdefault("flatProtLossOverTier", 0.10)
    armor.setdefault("flatProtLossUnderTier", 0.20)
    armor.setdefault("highDmgTierResistant", False)


# ================================================================
# PROJECTILE / BOW DAMAGE
# ================================================================

projectile_damage = {
    "crude arrow": {
        "Crude": 2.1,
        "Simple": 2.5,
        "Long": 3.0,
        "Cable Backed Long": 3.5,
        "Recurve": 4.0,
        "Composite": 4.5,
    },

    "flint/butcher arrow": {
        "Crude": 2.6,
        "Simple": 3.0,
        "Long": 3.5,
        "Cable Backed Long": 4.0,
        "Recurve": 4.5,
        "Composite": 5.0,
    },

    "obsidian arrow": {
        "Crude": 3.1,
        "Simple": 3.5,
        "Long": 4.0,
        "Cable Backed Long": 4.5,
        "Recurve": 5.0,
        "Composite": 5.5,
    },

    "copper arrow": {
        "Crude": 3.1,
        "Simple": 3.5,
        "Long": 4.0,
        "Cable Backed Long": 4.5,
        "Recurve": 5.0,
        "Composite": 5.5,
    },

    "bone arrow": {
        "Crude": 3.6,
        "Simple": 4.0,
        "Long": 4.5,
        "Cable Backed Long": 5.0,
        "Recurve": 5.5,
        "Composite": 6.0,
    },

    # "gold arrow": {
    #     "Crude": 4.1,
    #     "Simple": 4.5,
    #     "Long": 5.0,
    #     "Cable Backed Long": 5.5,
    #     "Recurve": 6.0,
    #     "Composite": 6.5,
    # },

    # "silver arrow": {
    #     "Crude": 4.1,
    #     "Simple": 4.5,
    #     "Long": 5.0,
    #     "Cable Backed Long": 5.5,
    #     "Recurve": 6.0,
    #     "Composite": 6.5,
    # },

    "SnBz arrow": {
        "Crude": 4.1,
        "Simple": 4.5,
        "Long": 5.0,
        "Cable Backed Long": 5.5,
        "Recurve": 6.0,
        "Composite": 6.5,
    },

    "BiBz arrow": {
        "Crude": 4.1,
        "Simple": 4.5,
        "Long": 5.0,
        "Cable Backed Long": 5.5,
        "Recurve": 6.0,
        "Composite": 6.5,
    },

    # "BlkBz arrow": {
    #     "Crude": 4.6,
    #     "Simple": 5.0,
    #     "Long": 5.5,
    #     "Cable Backed Long": 6.0,
    #     "Recurve": 6.5,
    #     "Composite": 7.0,
    # },

    "iron arrow": {
        "Crude": 5.6,
        "Simple": 6.0,
        "Long": 6.5,
        "Cable Backed Long": 7.0,
        "Recurve": 7.5,
        "Composite": 8.0,
    },

    # "meteoric arrow": {
    #     "Crude": 6.1,
    #     "Simple": 6.5,
    #     "Long": 7.0,
    #     "Cable Backed Long": 7.5,
    #     "Recurve": 8.0,
    #     "Composite": 8.5,
    # },

    "steel arrow": {
        "Crude": 6.6,
        "Simple": 7.0,
        "Long": 7.5,
        "Cable Backed Long": 8.0,
        "Recurve": 8.5,
        "Composite": 9.0,
    },
}


bow_types = [
    "Crude",
    "Simple",
    "Long",
    "Cable Backed Long",
    "Recurve",
    "Composite",
]


# ================================================================
# DAMAGE CALCULATION
# ================================================================

def calculate_damage(armor, damage, attack_tier):

    initial_damage = damage

    prot_tier = armor["protTier"]

    rel_prot = armor["relProt"]
    flat_prot = armor["flatProt"]

    high_dmg_tier_resistant = armor["highDmgTierResistant"]

    rel_loss_under = armor["relProtLossUnderTier"]
    rel_loss_over = armor["relProtLossOverTier"]

    flat_loss_under = armor["flatProtLossUnderTier"]
    flat_loss_over = armor["flatProtLossOverTier"]

    for tier in range(1, attack_tier + 1):

        above_tier = tier > prot_tier

        if above_tier:
            rel_loss = rel_loss_under
            flat_loss = flat_loss_under
        else:
            rel_loss = rel_loss_over
            flat_loss = flat_loss_over

        if above_tier and high_dmg_tier_resistant:
            rel_loss /= 2
            flat_loss /= 2

        rel_prot *= 1 - rel_loss
        flat_prot -= flat_loss

    damage = max(0, damage - flat_prot)

    damage *= 1 - max(0, rel_prot)

    damage_reduction = initial_damage - damage

    percent_reduction = damage_reduction / initial_damage

    return {
        "damage_taken": damage,
        "damage_reduction": damage_reduction,
        "percent_reduction": percent_reduction,
    }


# ================================================================
# BUILD ALL COMBINATIONS
# ================================================================

rows = []

for armor_name, armor in armor_stats.items():

    for projectile_name, bows in projectile_damage.items():

        for bow_name in bow_types:

            # ====================================================
            # ATTACK TIER COMES FROM THE BOW
            # ====================================================

            attack_tier = BOW_ATTACK_TIERS[bow_name]

            # Arrow + bow determine the initial damage
            base_damage = bows[bow_name]

            result = calculate_damage(
                armor=armor,
                damage=base_damage,
                attack_tier=attack_tier,
            )

            rows.append({
                "armor": armor_name,
                "projectile": projectile_name,
                "bow": bow_name,

                # This is now the BOW'S attack tier
                "attack_tier": attack_tier,

                "initial_damage": base_damage,
                "damage_taken": result["damage_taken"],
                "damage_reduction": result["damage_reduction"],
                "percent_reduction": result["percent_reduction"],
            })

df = pd.DataFrame(rows)

```

```python

# ================================================================
# CHECKBOXES
# ================================================================

armor_checks = {
    name: widgets.Checkbox(
        value=True,
        description=name,
        indent=False,
        layout=widgets.Layout(width="280px"),
    )
    for name in armor_stats
}

projectile_checks = {
    name: widgets.Checkbox(
        value=True,
        description=name,
        indent=False,
        layout=widgets.Layout(width="230px"),
    )
    for name in projectile_damage
}
# ================================================================
# PLOT
#
# X       = Armor Tier
# Y       = Damage Taken
# Color   = Armor
# Dash    = Bow
# Shape   = Projectile
# Jitter  = Armor model within tier
# ================================================================

import numpy as np

plot_output = widgets.Output()


# ================================================================
# FIXED ARMOR COLORS
#
# Each armor gets a permanent color. Selecting/deselecting armor
# will NOT change the colors or the legend.
# ================================================================

armor_palette = [
    "#1f77b4",  # blue
    "#ff7f0e",  # orange
    "#2ca02c",  # green
    "#d62728",  # red
    "#9467bd",  # purple
    "#8c564b",  # brown
    "#e377c2",  # pink
    "#7f7f7f",  # gray
    "#bcbd22",  # olive
    "#17becf",  # cyan
    "#393b79",
    "#637939",
    "#8c6d31",
    "#843c39",
    "#7b4173",
]

armor_names = list(armor_stats.keys())

armor_colors = {
    armor: armor_palette[i % len(armor_palette)]
    for i, armor in enumerate(armor_names)
}


# ================================================================
# STABLE JITTER
# ================================================================

armor_jitter = {}

for tier in sorted(
    set(armor["protTier"] for armor in armor_stats.values())
):

    tier_armors = [
        armor_name
        for armor_name, armor in armor_stats.items()
        if armor["protTier"] == tier
    ]

    offsets = np.linspace(
        -0.08,
        0.08,
        len(tier_armors),
    )

    for armor_name, offset in zip(
        tier_armors,
        offsets,
    ):
        armor_jitter[armor_name] = offset


# ================================================================
# SHARED ARMOR COLOR SCALE
#
# This EXACT scale is used by every layer.
# ================================================================

armor_color = alt.Color(
    "armor:N",
    title="Armor",
    scale=alt.Scale(
        domain=armor_names,
        range=[
            armor_colors[armor]
            for armor in armor_names
        ],
    ),
)


# ================================================================
# BOW DASH SCALE
# ================================================================

bow_dash = alt.StrokeDash(
    "bow:N",
    title="Bow",
    scale=alt.Scale(
        domain=bow_types,
        range=[
            [1, 0],          # Crude
            [6, 3],          # Simple
            [10, 4],         # Long
            [12, 4, 2, 4],   # Cable Backed Long
            [3, 3],          # Recurve
            [15, 5],         # Composite
        ],
    ),
)


# ================================================================
# PROJECTILE SHAPE
# ================================================================

projectile_shape = alt.Shape(
    "projectile:N",
    title="Projectile",
)


# ================================================================
# UPDATE PLOT
# ================================================================

def update_plot(*args):

    selected_armors = [
        name
        for name, checkbox in armor_checks.items()
        if checkbox.value
    ]

    selected_projectiles = [
        name
        for name, checkbox in projectile_checks.items()
        if checkbox.value
    ]

    with plot_output:

        clear_output(wait=True)

        if not selected_armors or not selected_projectiles:
            display(
                widgets.HTML(
                    "<b>Select at least one armor and one projectile.</b>"
                )
            )
            return

        # ------------------------------------------------------------
        # Filter
        # ------------------------------------------------------------

        filtered = df[
            df["armor"].isin(selected_armors)
            & df["projectile"].isin(selected_projectiles)
        ].copy()

        # ------------------------------------------------------------
        # Armor tier
        # ------------------------------------------------------------

        filtered["armor_tier"] = filtered["armor"].map(
            lambda armor: armor_stats[armor]["protTier"]
        )

        # ------------------------------------------------------------
        # Jittered x coordinate
        # ------------------------------------------------------------

        filtered["x_jitter"] = (
            filtered["armor_tier"]
            + filtered["armor"].map(armor_jitter)
        )

        # ============================================================
        # SHADED BOW DAMAGE RANGE
        # ============================================================

        band_df = (
            filtered
            .groupby(
                [
                    "armor",
                    "armor_tier",
                    "bow",
                ],
                as_index=False,
            )
            .agg(
                damage_min=("damage_taken", "min"),
                damage_max=("damage_taken", "max"),
            )
        )

        band_df["x_jitter"] = (
            band_df["armor_tier"]
            + band_df["armor"].map(armor_jitter)
        )

        bow_band = (
            alt.Chart(band_df)
            .mark_area(
                opacity=0.8,
            )
            .encode(
                x=alt.X(
                    "x_jitter:Q",
                    title="Armor Tier",
                    scale=alt.Scale(
                        domain=[
                            -0.5,
                            max(
                                armor["protTier"]
                                for armor in armor_stats.values()
                            ) + 0.5,
                        ]
                    ),
                    axis=alt.Axis(
                        values=sorted(
                            set(
                                armor["protTier"]
                                for armor in armor_stats.values()
                            )
                        )
                    ),
                ),

                y=alt.Y(
                    "damage_min:Q",
                    title="Damage Taken",
                    scale=alt.Scale(zero=True),
                ),

                y2="damage_max:Q",

                # IMPORTANT:
                # Same armor color scale as points and lines.
                color=armor_color,

                detail=[
                    "armor:N",
                    "bow:N",
                ],
            )
        )

        # ============================================================
        # LINES
        # ============================================================

        lines = (
            alt.Chart(filtered)
            .mark_line(
                strokeWidth=2,
                opacity=0.75,
            )
            .encode(
                x=alt.X(
                    "x_jitter:Q",
                    title="Armor Tier",
                    axis=alt.Axis(
                        values=sorted(
                            set(
                                armor["protTier"]
                                for armor in armor_stats.values()
                            )
                        )
                    ),
                ),

                y=alt.Y(
                    "damage_taken:Q",
                    title="Damage Taken",
                    scale=alt.Scale(zero=True),
                ),

                # ARMOR = COLOR
                color=armor_color,

                # BOW = DASH
                strokeDash=bow_dash,

                detail=[
                    "armor:N",
                    "bow:N",
                    "projectile:N",
                ],
            )
        )

        # ============================================================
        # POINTS
        # ============================================================

        points = (
            alt.Chart(filtered)
            .mark_point(
                filled=True,
                size=90,
                opacity=0.95,
            )
            .encode(
                x=alt.X(
                    "x_jitter:Q",
                    title="Armor Tier",
                    axis=alt.Axis(
                        values=sorted(
                            set(
                                armor["protTier"]
                                for armor in armor_stats.values()
                            )
                        )
                    ),
                ),

                y=alt.Y(
                    "damage_taken:Q",
                    title="Damage Taken",
                    scale=alt.Scale(zero=True),
                ),

                # ARMOR = COLOR
                color=armor_color,

                # PROJECTILE = SHAPE
                shape=projectile_shape,

                tooltip=[
                    alt.Tooltip(
                        "armor:N",
                        title="Armor",
                    ),
                    alt.Tooltip(
                        "armor_tier:Q",
                        title="Armor Tier",
                    ),
                    alt.Tooltip(
                        "bow:N",
                        title="Bow",
                    ),
                    alt.Tooltip(
                        "projectile:N",
                        title="Projectile",
                    ),
                    alt.Tooltip(
                        "attack_tier:Q",
    title="Bow Attack Tier",
                    ),
                    alt.Tooltip(
                        "initial_damage:Q",
                        title="Bow + Arrow Damage",
                        format=".2f",
                    ),
                    alt.Tooltip(
                        "damage_taken:Q",
                        title="Damage Taken",
                        format=".2f",
                    ),
                    alt.Tooltip(
                        "damage_reduction:Q",
                        title="Damage Prevented",
                        format=".2f",
                    ),
                    alt.Tooltip(
                        "percent_reduction:Q",
                        title="Damage Reduction",
                        format=".2%",
                    ),
                ],
            )
        )

        # ============================================================
        # COMBINE
        # ============================================================

        chart = (
            (bow_band + lines + points)
            .resolve_scale(
                color="shared",
            )
            .properties(
                width=1100,
                height=650,
                title="Damage Taken by Armor Tier",
            )
            .interactive()
        )

        display(chart)


# ================================================================
# CALLBACKS
# ================================================================

for checkbox in armor_checks.values():
    checkbox.observe(update_plot, names="value")

for checkbox in projectile_checks.values():
    checkbox.observe(update_plot, names="value")


# ================================================================
# CONTROLS
# ================================================================

armor_box = widgets.VBox(
    [
        widgets.HTML("<h3>Armor</h3>"),
        *armor_checks.values(),
    ],
    layout=widgets.Layout(
        width="310px",
        max_height="700px",
        overflow="auto",
    ),
)

projectile_box = widgets.VBox(
    [
        widgets.HTML("<h3>Projectile</h3>"),
        *projectile_checks.values(),
    ],
    layout=widgets.Layout(
        width="280px",
        max_height="700px",
        overflow="auto",
    ),
)

controls = widgets.HBox(
    [
        armor_box,
        projectile_box,
    ],
    layout=widgets.Layout(
        align_items="flex-start",
    ),
)


# ================================================================
# DISPLAY
# ================================================================

display(controls)
display(plot_output)

update_plot()
```
